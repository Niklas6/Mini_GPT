```python
with open('tinystories.txt', 'r', encoding='utf-8') as f:
    text = f.read()
```

```python
tokenizer = Tokenizer(
    BPE(unk_token="<unk>")
)



tokenizer.pre_tokenizer = ByteLevel(
    add_prefix_space=False
)
tokenizer.decoder = ByteLevelDecoder()
trainer = BpeTrainer(
    vocab_size=1024,
    min_frequency=2,
    special_tokens=["<unk>"],
    #initial_alphabet=ByteLevel.alphabet(),
)

tokenizer.train(
    ["tinystories.txt"],
    trainer,
)

#tokenizer.save("tokenizer.json")

```

```python

E = [
    "Once upon a time there was a little girl.",
    "The dragon couldn't find its family!",
    "She carefully opened the mysterious box.",
    "Hello! £5, café, naïve, 2026—really?",
]
for e in E:
    e2=tokenizer.encode(e)
    print(e)
    print(e2.tokens)
    print(e2.ids)
    print(tokenizer.decode(e2.ids))
```
