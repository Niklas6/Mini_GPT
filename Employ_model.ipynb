```python
import torch
import Train_model as llm

# Select device


# Load checkpoint onto CPU first for portability
checkpoint = torch.load(
    "tiny_transformer.pt",
    map_location="cpu",
)

chars = checkpoint["chars"]
config = checkpoint["config"]

model = llm.BigramLanguageModel()
model.load_state_dict(checkpoint["model_state_dict"])

model.eval()


stoi = {character: index for index, character in enumerate(chars)}
itos = {index: character for index, character in enumerate(chars)}

encode = lambda s: [stoi[c] for c in s]
decode = lambda l: ''.join([itos[i] for i in l])

# Generate text
prompt = "Once upon a time"

input_tokens = torch.tensor(
    [encode(prompt)],
    dtype=torch.long,
)

gen = model.generate(
        input_tokens,
        max_new_tokens=500, )
print(decode(gen[0].tolist()))
```
